```python
### Improved ATM


bal = 10000
pin = 1234

p = int(input("Enter Pin: "))

if p == pin:

    while True:
        print("\n1. Balance")
        print("2. Withdraw")
        print("3. Deposit")
        print("4. Exit")

        a = int(input("Enter Choice: "))

        if a == 1:
            print(f"Balance: {bal}")

        elif a == 2:
            b = int(input("Enter Amount: "))

            if b <= bal:
                bal = bal - b
                print(f"Remaining Balance: {bal}")
            else:
                print("Insufficient Balance")

        elif a == 3:
            d = int(input("Enter Amount: "))
            bal = bal + d
            print(f"Remaining Balance: {bal}")

        elif a == 4:
            print("Thank you for using our ATM.")
            break

        else:
            print("Invalid Choice")

else:
    print("Wrong Pin")
```
