#### `while` Loops

```python
# Print numbers from 1 to 10 using while

i = 1

while i <= 10:
    print(i)
    i += 1
```

```python
# Print numbers from 10 to 1 using while

i = 10

while i >= 1:
    print(i)
    i -= 1
```

```python
# Print all multiples of 3 till 50

i = 3

while i <= 50:
    print(i)
    i += 3
```

```python
# Take a number from user and print its multiplication table

n = int(input("Enter a number: "))

i = 1

while i <= 10:
    print(n, "x", i, "=", n * i)
    i += 1
```

```python
# Count digits in a number

n = int(input())

if n == 0:
    print(1)
else:
    count = 0

    while n > 0:
        count += 1
        n //= 10

    print(count)
```

```python
# Reverse a number

n = int(input("Enter a number: "))

rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10

print(rev)
```

```python
# Check whether a number is palindrome

n = int(input("Enter a number: "))

original = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10

if original == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Find sum of digits

n = int(input("Enter a number: "))

total = 0

while n > 0:
    digit = n % 10
    total += digit
    n //= 10

print(total)
```

```python
# Find product of digits

n = int(input("Enter a number: "))

product = 1

while n > 0:
    digit = n % 10
    product *= digit
    n //= 10

print(product)
```

#### Loops with Strings

```python
# Print each character of a string separately.

s="Hello"

for ch in s:
    print(ch)
```

```python
# Count vowels in a string.

s="Apollo"
count=0

for ch in s:
    if ch in "aeiouAEIOU":
        count+=1
print(count)
```

```python
# Count uppercase and lowercase letters.

s = "Hello This Is UPPERCASE and lowercase Sentence"

count_upper = 0
count_lower = 0

for ch in s:
    if ch.isupper():
        count_upper += 1
    elif ch.islower():
        count_lower += 1

print(f"UpperCase letters : {count_upper}")
print(f"LowerCase letters : {count_lower}")
```

```python
# Reverse a string using loop.

s="Hello"
reverse=""

for ch in s:
    reverse=ch + reverse
print(reverse)

```

```python
# Check whether string is palindrome.

s="madam"
rev=""

for ch in s:
    rev=ch+rev
    
if rev==s:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Count frequency of a character.

s="bananA"
count=0

for ch in s:
    if ch=="a" or ch=="A":
        count+=1
print(f"{s},a->{count}")
```

```python
# Remove all spaces from string.

s="My Name is Alice"
N=""

for ch in s:
    if ch !=" ":
        N=N+ch
print(N)
```

```python
# Print only vowels from a string.

s="Apollo"

for ch in s:
    if ch in "aeiouAEIOU":
        print(ch)
```

```python
# Find longest word in a sentence.

s="This is Python Programming Practice Module"

word=s.split()
longest=word[0]

for ch in word:
    if len(ch) > len(longest):
        longest=ch
print(longest)
```

```python
# Convert lowercase letters to uppercase and viceversa manually using loops.

s = "Hello Python"

result = ""

for ch in s:
    if ch.islower():
        result += ch.upper()
    elif ch.isupper():
        result += ch.lower()
    else:
        result += ch

print(result)
```

#### Nested Loops

```python
n=5
for i in range(n):
    for j in range(i+1):
        print("@",end=" ")
    print("")
```

```python
n=5
for i in range(n):
    for j in range(i,n):
        print("@",end=" ")
    print("")
```

```python
n=5
for i in range(n):
    for j in range(i,n):
        print("",end=" ")
    for j in range(i+1):
        print("@",end=" ")
    print("")
```

```python
# Print right pyramid.
n=5
for i in range(n):
    for j in range(i,n):
        print(" ",end=" ")
    for j in range(i):
        print("@",end=" ")
    for j in range(i+1):
        print("@",end=" ")
    print("")
```

```python
for i in range(1,11):
    n = int(input("Enter a number: "))

i = 1

while i <= 10:
    print(n, "x", i, "=", n * i)
    i += 1
```

```python
n=1
while n<=10:
    print(f"\nTable of {n}")

    for i in range(1,11):
        print(f"{n} x {i} = {n*i}")
    n+=1
```

```python
# Print chessboard pattern using loops.

for row in range(8):
    for col in range(8):
        if (row + col) % 2 == 0:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Create a number triangle.

for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=" ")
    print("")
```

```python
# Another number triangle.

for i in range(1, 6):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
# Print Floyd’s triangle.
m=1
for i in range(1,10):
    for j in range(i):
        print(m,end=" ")
        m+=1
    print("")
```

```python
# Print diamond pattern.

n=5
for i in range(n):
    for j in range(i,n):
        print(" ",end=" ")
    for j in range(i):
        print("@",end=" ")
    for j in range(i+1):
        print("@",end=" ")
    print("")
        
n=5
for i in range(1,n):
    for j in range(i+1):
        print(" ",end=" ")
    for j in range(i,n):
        print("@",end=" ")
    for j in range(i,n-1):
        print("@",end=" ")
    print("")
```
