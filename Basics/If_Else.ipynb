### Age Group Categorization

```python
age=int(input("Enter Your Age"))

if age<13:
    print("You are a Child")
elif age<20:
    print("You are Teenager")
elif age <60:
    print("You are Adult")
else:
    print("You are Senior")
    
    # Age=20
```

#### Determine if fruit is ripe,overripe or unripe based on its color

```python
fruit = input("Enter fruit name: ").strip().lower()
color = input("Enter fruit color: ").strip().lower()

if fruit == "banana":
    if color == "yellow":
        print("Ripe")
    elif color == "green":
        print("Unripe")
    elif color == "brown":
        print("Overripe")
    else:
        print("Invalid color for banana.")

elif fruit == "apple":
    if color == "green":
        print("Unripe")
    elif color == "red":
        print("Ripe")
    else:
        print("Invalid color for apple.")

elif fruit == "mango":
    if color == "green":
        print("Unripe")
    elif color == "yellow":
        print("Ripe")
    else:
        print("Invalid color for mango.")

else:
    print("Sorry, this fruit is not supported.")
```

### Weather Activity Suggestion

```python
weather = input("Enter the weather (sunny, rainy, snowy, cloudy): ").strip().lower()

if weather == "sunny":
    print("Go for a walk or play outside.")

elif weather == "rainy":
    print("Stay indoors and read a book.")

elif weather == "snowy":
    print("Build a snowman or go skiing.")

elif weather == "cloudy":
    print("Go for a short walk or visit a cafe.")

else:
    print("Sorry, I don't have suggestions for this weather.")
    
    #sunny
```

### Password Checker



```python
a = input("Enter Your Password to Check Strength: ")

if " " in a:
    print("Password should not contain spaces.")

elif len(a) < 6:
    print("Weak Password")

elif len(a) >= 10 and "@" in a and any(i.isdigit() for i in a):
    print("Strong Password")

elif len(a) >= 8 and "@" in a:
    print("Medium Password")

else:
    print("Weak Password")
    
    #Hello@123=>weak
    #HelloPython@12345=>strong
    
```

### Leap year Checker

```python
year = 2025

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print( year, " is a leap year")
else:
    print(year, "is NOT a leap year")
```
