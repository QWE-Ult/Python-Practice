#### Polymorphism func - Write a func that multiples two numbers, but can also accept and multiply strings.

```python
def multiply(a,b):
    return a*b


x=multiply(2,3)
z=multiply(5,"c")

print(x)
print(z)
```

#### Create a func that returns both the area and circumferenc of a circle given its radius

```python
def info(r):
    area=3.14*r*r
    circumference=2*3.14*r
    # return ("area of circle is : ",area,"circumference of circle is : ",circumference)
    return area,circumference

a,c=info(3)
# print("Area : ",a, "Circumference : ",c)
print(f"Area: {a:.2f}, Circumference: {c:.2f}")
```

#### Wrie a function that greets a user, if no name , it should greet with default

#### Func with *args

```python
def sum_all(*args):
    return sum(args)

print(sum_all(1,2))
print(sum_all(1,2,3,4,5,6))
```

#### func with **kwargs

```python
def a_kwargs(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
        
        
a_kwargs(name="a",age=22,role="lifter")
```

#### Recursive function

```python
def fact(n):
    if n==0:
        return 1
    else :
        return n* fact(n-1)
    
fact(5)
```
